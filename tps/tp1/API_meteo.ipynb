```python
import json
import requests
```

Coordonnées geographiques de Bruxelles et une date choisie parmi les 51 fichiers du corpus CAmille.

```python

latitude = 50.85

longitude = 4.35

date_journal = "1940-03-01"
```

Parametre de requete API 

```python
url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude" : latitude,
    "longitude" : longitude,
    "start_date" : date_journal,
    "end_date" : date_journal,
    "hourly" : "temperature_2m,rain"

}

response = requests.get(url, params=params)
weather_data = response.json()
```

Afficher la météo (matin, midi et soir), la température ainsi que la quantité de pluie

```python
try:

    hourly_times = weather_data ["hourly"]["time"]
    hourly_temps = weather_data ["hourly"]["temperature_2m"]
    hourly_rain = weather_data ["hourly"]["rain"]

    print(f"Données météo dans le corpus CAmille:")

    print(f"Bruxelles, le journal du : {date_journal}\n")

    
    print(f"À {hourly_times [8][-5:]} (Matin)| Température = {hourly_temps [8]} °C | pluie: {hourly_rain [8]} mm")
        
    print(f"À {hourly_times [13][-5:]} (Midi)| Température = {hourly_temps [13]} °C | pluie: {hourly_rain [13]} mm")

    print(f"À {hourly_times [20][-5:]} (Soir)| Température = {hourly_temps [20]} °C | pluie: {hourly_rain [20]} mm")

# -5 afin d'adopter le format d'heure 00:00 

except KeyError:
    print(f"Données météo non trouvées")
```
